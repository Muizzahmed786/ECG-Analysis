```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Basic Dataset information (rows, columns, attribute names, attribute type)

df = pd.read_csv("./data/AttributesDictionary.csv")

rows = df.shape[0]
columns = df.shape[1]

attribute_names = df.columns.tolist()
print(attribute_names)
print("\n")
df.info()
```

```python
# Checking for duplicates in suspected columns
check = ["Filename", "ECG_ID"]

for attribute in check:
    counts = df[attribute].value_counts()
    duplicates = counts[counts > 1]
    
    if not duplicates.empty:
        print(f"Duplicates found in '{attribute}'")
        
        for value, count in duplicates.items():
            print(f"Value '{value}' appears {count} times")
        print()
    else:
        print(f"No duplicates found in '{attribute}")
```

```python
# NULL value counts

null_counts = df.isna().sum()
print(null_counts)
```

```python
# Analasis of Patient data

unique_patients = set(df['Patient_ID'])
print(f"No. of unique patients : {len(unique_patients)}")
print(f"Total patients : {len(df['Patient_ID'])}")

# Strip of 'd' from age
df['Age'] = df['Age'].str.replace('d', '').astype(int)

min_age = df['Age'].min()
max_age = df['Age'].max()

print(f"Minimum Age (in days) : {min_age}")
print(f"Maximum Age (in days) : {max_age}")
```

```python
# Gender spread

print(f"{df.Gender.value_counts(normalize=True).map(lambda x: f'{x*100:.2f}%')}")
```
